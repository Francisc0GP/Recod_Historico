```python
def DenominadoresSustancias(df):
#     sustancias_legales = ['Tabaco', 'Alcohol']
#     sustancias_ilegales = [
#     "Marihuana",
#     "Hachis",
#     "Cocaína",
#     "Crack",
#     "Otras Presentaciones (Basuco o pasta base, cocaina negra)",
#     "Otros (aire comprimido, gasolinas y combustibles)",
#     "Solventes y removedores",
#     "Pegamento",
#     "Esmaltes y pinturas",
#     "Otros (derivados anfetaminicos)",
#     "Anfetaminas",
#     "Otras (sedantes hiptnoticos, GHB)",
#     "Metanfetaminas",
#     "MDMA(extasis) y metanfetaminas alucinogenas (DMT)",
#     "Otras Sustancias",
#     "Otras (PCP, ketamina, excepto metanfetamina)",
#     "Plantas alucinogenas y derivados",
#     "LSD",
#     "Benzodiazepinas",
#     "Rohypnol",
#     "Opio y opiodes (morfina, codeina)",
#     "Heroina",
#     "Opiaceos sinteticos (propoxifeno, nailbufina)",
#     "Con utilidad medica (Prozac, Paxil, Carbamazepina)"
# ]
#     df_legales = df[df['DrogaImpacto'].isin(sustancias_legales)]
#     violaciones = df_legales[sustancias_ilegales].ne(0).any(axis=1)
#     df_sinconsumoilegales = df_legales[~violaciones]
#     df_EceRiegsIlegales = df[~df['FolioId'].isin(df_sinconsumoilegales['FolioId'])]
    df_EceRiegsIlegales = df[df['caso'] == 1]
    return df_EceRiegsIlegales

def list_cols_Sust():   
    list_cols = [
        'TabacoUM', 'AlcoholUM', 'MarihuanaUM', 'HachisUM', 'CocaínaUM',
        'CrackUM', 'Otras Presentaciones (Basuco o pasta base, cocaina negra)UM', 'Otros (aire comprimido, gasolinas y combustibles)UM',
        'Solventes y removedoresUM', 'PegamentoUM', 'Esmaltes y pinturasUM', 'Otros (derivados anfetaminicos)UM', 'AnfetaminasUM',
        'Otras (sedantes hiptnoticos, GHB)UM', 'MetanfetaminasUM', 'MDMA(extasis) y metanfetaminas alucinogenas (DMT)UM',
        'Otras SustanciasUM', 'Otras (PCP, ketamina, excepto metanfetamina)UM', 'Plantas alucinogenas y derivadosUM', 'LSDUM',
        'BenzodiazepinasUM', 'RohypnolUM', 'Opio y opiodes (morfina, codeina)UM', 'HeroinaUM', 'Opiaceos sinteticos (propoxifeno, nailbufina)UM',   
        'Con utilidad medica (Prozac, Paxil, Carbamazepina)UM']
    list_cols_sin_um = [
        'Tabaco', 'Alcohol', 'Marihuana', 'Hachis', 'Cocaína',
        'Crack', 'Otras Presentaciones (Basuco o pasta base, cocaina negra)', 'Otros (aire comprimido, gasolinas y combustibles)',
        'Solventes y removedores', 'Pegamento', 'Esmaltes y pinturas', 'Otros (derivados anfetaminicos)', 'Anfetaminas',
        'Otras (sedantes hiptnoticos, GHB)', 'Metanfetaminas', 'MDMA(extasis) y metanfetaminas alucinogenas (DMT)',
        'Otras Sustancias', 'Otras (PCP, ketamina, excepto metanfetamina)', 'Plantas alucinogenas y derivados', 'LSD',
        'Benzodiazepinas', 'Rohypnol', 'Opio y opiodes (morfina, codeina)', 'Heroina', 'Opiaceos sinteticos (propoxifeno, nailbufina)',
        'Con utilidad medica (Prozac, Paxil, Carbamazepina)']
    return list_cols, list_cols_sin_um

def list_cols_Grupo():
    cols_grupos = ['Alcohol','Otras Sustancias','Tabaco','Marihuana','Cocaína','Inhalables','Metanfetaminas','Alucinógenos','Medicamentos','Opioides']
    cols_gruposUM = ['AlcoholUM','Otras SustanciasUM','TabacoUM','MarihuanaUM','CocaínaUM','InhalablesUM','MetanfetaminasUM','AlucinógenosUM','MedicamentosUM','OpioidesUM']
    return cols_grupos, cols_gruposUM

def dict_grupos():
    return {
        'Tabaco' : 'Tabaco',
        'Alcohol' : 'Alcohol',
        'Marihuana' : 'Marihuana',
        'Hachis' : 'Marihuana',
        'Cocaina' : 'Cocaína',
        'Crack' : 'Cocaína',
        'Otras Presentaciones (Basuco o pasta base, cocaina negra)' : 'Cocaína',
        'Solventes y removedores' : 'Inhalables',
        'Pegamento' : 'Inhalables',
        'Esmaltes y pinturas' : 'Inhalables',
        'Otros (aire comprimido, gasolinas y combustibles)' : 'Inhalables',
        'Anfetaminas': 'Metanfetaminas',
        'Metanfetaminas' : 'Metanfetaminas',
        'MDMA(extasis) y metanfetaminas alucinogenas (DMT)' : 'Metanfetaminas',
        'Otros (derivados anfetaminicos)' : 'Metanfetaminas',
        'LSD' : 'Alucinógenos',
        'Plantas alucinogenas y derivados' : 'Alucinógenos',
        'Otras (PCP, ketamina, excepto metanfetamina)' : 'Alucinógenos',
        'Benzodiazepinas' : 'Medicamentos',
        'Rohypnol' : 'Medicamentos',
        'Otras (sedantes hiptnoticos, GHB)' : 'Medicamentos',
        'Heroina' : 'Opioides',
        'Opiaceos sinteticos (propoxifeno, nailbufina)' : 'Opioides',
        'Opio y opiodes (morfina, codeina)' : 'Opioides',
        'Con utilidad medica (Prozac, Paxil, Carbamazepina)' : 'Medicamentos',
        'Otras Sustancias' : 'Otras Sustancias'
    }

def safe_convert_to_float(value):
    try:
        return float(value)
    except ValueError:
        return None

def Set_Grupo(df):
    dfgrupos = df.copy() # DataFrame Para crear los grupos de drogas
    dfgrupos['DrogaImpacto'] = dfgrupos['DrogaImpacto'].map(dict_grupos())
    dfgrupos['Marihuana'] = dfgrupos[['Marihuana', 'Hachis']].max(axis=1)
    dfgrupos['Cocaína'] = dfgrupos[['Cocaína', 'Crack', 'Otras Presentaciones (Basuco o pasta base, cocaina negra)']].max(axis=1)
    dfgrupos['Inhalables'] = dfgrupos[['Solventes y removedores', 'Pegamento', 'Esmaltes y pinturas', 'Otros (aire comprimido, gasolinas y combustibles)']].max(axis=1)
    dfgrupos['Metanfetaminas'] = dfgrupos[['Anfetaminas', 'Metanfetaminas', 'MDMA(extasis) y metanfetaminas alucinogenas (DMT)', 'Otros (derivados anfetaminicos)']].max(axis=1)
    dfgrupos['Alucinógenos'] = dfgrupos[['LSD', 'Plantas alucinogenas y derivados', 'Otras (PCP, ketamina, excepto metanfetamina)']].max(axis=1)
    dfgrupos['Medicamentos'] = dfgrupos[['Benzodiazepinas', 'Rohypnol', 'Otras (sedantes hiptnoticos, GHB)', 'Con utilidad medica (Prozac, Paxil, Carbamazepina)']].max(axis=1)
    dfgrupos['Opioides'] = dfgrupos[['Heroina', 'Opiaceos sinteticos (propoxifeno, nailbufina)', 'Opio y opiodes (morfina, codeina)']].max(axis=1)
    cols_derecha_no_izq = [
    'Hachis','Crack','Otras Presentaciones (Basuco o pasta base, cocaina negra)','Solventes y removedores','Pegamento','Esmaltes y pinturas','Otros (aire comprimido, gasolinas y combustibles)',
    'Anfetaminas','MDMA(extasis) y metanfetaminas alucinogenas (DMT)','Otros (derivados anfetaminicos)','LSD','Plantas alucinogenas y derivados',
    'Otras (PCP, ketamina, excepto metanfetamina)','Benzodiazepinas','Rohypnol','Otras (sedantes hiptnoticos, GHB)','Con utilidad medica (Prozac, Paxil, Carbamazepina)','Heroina','Opiaceos sinteticos (propoxifeno, nailbufina)','Opio y opiodes (morfina, codeina)']
    dfgrupos.drop(columns=cols_derecha_no_izq, inplace=True)

    dfgrupos['MarihuanaUM'] = dfgrupos[['MarihuanaUM', 'HachisUM']].max(axis=1)
    dfgrupos['CocaínaUM'] = dfgrupos[['CocaínaUM', 'CrackUM', 'Otras Presentaciones (Basuco o pasta base, cocaina negra)UM']].max(axis=1)
    dfgrupos['InhalablesUM'] = dfgrupos[['Solventes y removedoresUM', 'PegamentoUM', 'Esmaltes y pinturasUM', 'Otros (aire comprimido, gasolinas y combustibles)UM']].max(axis=1)
    dfgrupos['MetanfetaminasUM'] = dfgrupos[['AnfetaminasUM', 'MetanfetaminasUM', 'MDMA(extasis) y metanfetaminas alucinogenas (DMT)UM', 'Otros (derivados anfetaminicos)UM']].max(axis=1)
    dfgrupos['AlucinógenosUM'] = dfgrupos[['LSDUM', 'Plantas alucinogenas y derivadosUM', 'Otras (PCP, ketamina, excepto metanfetamina)UM']].max(axis=1)
    dfgrupos['MedicamentosUM'] = dfgrupos[['BenzodiazepinasUM', 'RohypnolUM', 'Otras (sedantes hiptnoticos, GHB)UM', 'Con utilidad medica (Prozac, Paxil, Carbamazepina)UM']].max(axis=1)
    dfgrupos['OpioidesUM'] = dfgrupos[['HeroinaUM', 'Opiaceos sinteticos (propoxifeno, nailbufina)UM', 'Opio y opiodes (morfina, codeina)UM']].max(axis=1)
    cols_derecha_no_izqUM = [
    'HachisUM','CrackUM','Otras Presentaciones (Basuco o pasta base, cocaina negra)UM','Solventes y removedoresUM','PegamentoUM','Esmaltes y pinturasUM','Otros (aire comprimido, gasolinas y combustibles)UM',
    'AnfetaminasUM','MDMA(extasis) y metanfetaminas alucinogenas (DMT)UM','Otros (derivados anfetaminicos)UM','LSDUM','Plantas alucinogenas y derivadosUM',
    'Otras (PCP, ketamina, excepto metanfetamina)UM','BenzodiazepinasUM','RohypnolUM','Otras (sedantes hiptnoticos, GHB)UM','Con utilidad medica (Prozac, Paxil, Carbamazepina)UM','HeroinaUM','Opiaceos sinteticos (propoxifeno, nailbufina)UM','Opio y opiodes (morfina, codeina)UM']
    dfgrupos.drop(columns=cols_derecha_no_izqUM, inplace=True)

    cols_derecha_EdadInicio = [
    # Marihuana
    'EdadInicioMarihuana', 'EdadInicioHachis',

    # Cocaína
    'EdadInicioCocaína', 'EdadInicioCrack',
    'EdadInicioOtras Presentaciones (Basuco o pasta base, cocaina negra)',

    # Inhalables
    'EdadInicioSolventes y removedores', 'EdadInicioPegamento',
    'EdadInicioEsmaltes y pinturas',
    'EdadInicioOtros (aire comprimido, gasolinas y combustibles)',

    # Metanfetaminas
    'EdadInicioAnfetaminas', 'EdadInicioMetanfetaminas',
    'EdadInicioMDMA(extasis) y metanfetaminas alucinogenas (DMT)',
    'EdadInicioOtros (derivados anfetaminicos)',

    # Alucinógenos
    'EdadInicioLSD', 'EdadInicioPlantas alucinogenas y derivados',
    'EdadInicioOtras (PCP, ketamina, excepto metanfetamina)',

    # Medicamentos
    'EdadInicioBenzodiazepinas', 'EdadInicioRohypnol',
    'EdadInicioOtras (sedantes hiptnoticos, GHB)',
    'EdadInicioCon utilidad medica (Prozac, Paxil, Carbamazepina)',

    # Opioides
    'EdadInicioHeroina',
    'EdadInicioOpiaceos sinteticos (propoxifeno, nailbufina)',
    'EdadInicioOpio y opiodes (morfina, codeina)'
    ]

    for col in cols_derecha_EdadInicio:
        dfgrupos[col] = dfgrupos[col].apply(safe_convert_to_float)
    # Agrupación por edad de inicio
    dfgrupos['EdadInicioMarihuana'] = dfgrupos[['EdadInicioMarihuana', 'EdadInicioHachis']].min(axis=1)
    dfgrupos['EdadInicioCocaína'] = dfgrupos[['EdadInicioCocaína', 'EdadInicioCrack', 'EdadInicioOtras Presentaciones (Basuco o pasta base, cocaina negra)']].min(axis=1)
    dfgrupos['EdadInicioInhalables'] = dfgrupos[['EdadInicioSolventes y removedores', 'EdadInicioPegamento', 'EdadInicioEsmaltes y pinturas', 'EdadInicioOtros (aire comprimido, gasolinas y combustibles)']].min(axis=1)
    dfgrupos['EdadInicioMetanfetaminas'] = dfgrupos[['EdadInicioAnfetaminas', 'EdadInicioMetanfetaminas', 'EdadInicioMDMA(extasis) y metanfetaminas alucinogenas (DMT)', 'EdadInicioOtros (derivados anfetaminicos)']].min(axis=1)
    dfgrupos['EdadInicioAlucinógenos'] = dfgrupos[['EdadInicioLSD', 'EdadInicioPlantas alucinogenas y derivados', 'EdadInicioOtras (PCP, ketamina, excepto metanfetamina)']].min(axis=1)
    dfgrupos['EdadInicioMedicamentos'] = dfgrupos[['EdadInicioBenzodiazepinas', 'EdadInicioRohypnol', 'EdadInicioOtras (sedantes hiptnoticos, GHB)', 'EdadInicioCon utilidad medica (Prozac, Paxil, Carbamazepina)']].min(axis=1)
    dfgrupos['EdadInicioOpioides'] = dfgrupos[['EdadInicioHeroina', 'EdadInicioOpiaceos sinteticos (propoxifeno, nailbufina)', 'EdadInicioOpio y opiodes (morfina, codeina)']].min(axis=1)


    # Columnas a eliminar (las que ya se agruparon)
    cols_edad_no_finales = [
        'EdadInicioHachis',
        'EdadInicioCrack',
        'EdadInicioOtras Presentaciones (Basuco o pasta base, cocaina negra)',
        'EdadInicioSolventes y removedores',
        'EdadInicioPegamento',
        'EdadInicioEsmaltes y pinturas',
        'EdadInicioOtros (aire comprimido, gasolinas y combustibles)',
        'EdadInicioAnfetaminas',
        'EdadInicioMDMA(extasis) y metanfetaminas alucinogenas (DMT)',
        'EdadInicioOtros (derivados anfetaminicos)',

        'EdadInicioLSD',
        'EdadInicioPlantas alucinogenas y derivados',
        'EdadInicioOtras (PCP, ketamina, excepto metanfetamina)',
        'EdadInicioBenzodiazepinas',
        'EdadInicioRohypnol',
        'EdadInicioOtras (sedantes hiptnoticos, GHB)',
        'EdadInicioCon utilidad medica (Prozac, Paxil, Carbamazepina)',
        'EdadInicioHeroina',
        'EdadInicioOpiaceos sinteticos (propoxifeno, nailbufina)',
        'EdadInicioOpio y opiodes (morfina, codeina)'
    ]


    dfgrupos.drop(columns=cols_edad_no_finales, inplace=True)

    return dfgrupos

def Merge_dataframes(df1, df2, df3):
    df = pd.concat([df1, df2, df3], ignore_index=True)
    print (f'Tamaño dataframe merge 3 primeros df : {len(df)}')
    df['FolioId'] = df['FolioId'].astype(str) + '-' + df['Semestre'].astype(str)
    df.sort_values('Semestre', inplace=True)
    # df.drop_duplicates(subset='FolioId', keep='last', inplace=True)
    # Prueba para generar denominadores antes
    dfIlegales = DenominadoresSustancias(df.copy())
    dfgruposIlegales = Set_Grupo(dfIlegales.copy())
    print (f'Tamaño dataframe Ilegales : {len(dfIlegales)}', f'Tamaño dataframe Ilegales Grupos : {len(dfgruposIlegales)}')
    dfgrupos = Set_Grupo(df.copy())
    print (f'Tamaño dataframe Grupos : {len(dfgrupos)}')
    dfsust = df.copy()
    return dfgrupos, dfsust, dfgruposIlegales, dfIlegales

def Read_data(file_path):
    df1 = pd.read_csv(f'{file_path}\\data\\Db_2004_2009.csv')
    df2 = pd.read_csv(f'{file_path}\\data\\db_2010.csv')
    df3 = pd.read_csv(f'{file_path}\\data\\db_2011_13.csv')
    dfgrupos, dfsust , dfgruposIlegales, dfsustanciasIlegales = Merge_dataframes(df1, df2, df3)
    dfgrupos.drop(columns=['caso'], inplace=True)
    dfsust.drop(columns=['caso'], inplace=True)
    dfgruposIlegales.drop(columns=['caso'], inplace=True)
    dfsustanciasIlegales.drop(columns=['caso'], inplace=True)
    return dfgrupos, dfsust, dfgruposIlegales, dfsustanciasIlegales

def DrogaImpacto(dfgrupo, dfsust):
    list_cols, list_cols_sin_um = list_cols_Grupo()
    dfgrupo.drop(columns=list_cols + list_cols_sin_um, inplace=True)
    dummies = pd.get_dummies(dfgrupo['DrogaImpacto'])
    dfgrupo = pd.concat([dfgrupo, dummies], axis=1)
    print (f'DrogaImp Tamaño dataframe Grupos : {len(dfgrupo)}')
    dfgrupo.to_csv(f'{os.getcwd()}\\result\\EI_DROGAIMP_2004_2013(GrupoLegalesIlegales).csv', index=False)
    # Sustancias
    list_cols, list_cols_sin_um = list_cols_Sust()
    dfsust.drop(columns=list_cols + list_cols_sin_um, inplace=True)
    dummies = pd.get_dummies(dfsust['DrogaImpacto'])
    dfsust = pd.concat([dfsust, dummies], axis=1)
    print (f'DrogaImp Tamaño dataframe Sustancias : {len(dfsust)}')
    dfsust.to_csv(f'{os.getcwd()}\\result\\EI_DROGAIMP_2004_2013(SustanciasLegalesIlegales).csv', index=False)
    return dfgrupo, dfsust

def AlgunaVez(dfgrupo, dfsust):
    list_cols, list_cols_sin_um = list_cols_Grupo()
    dfgrupo.drop(columns=list_cols_sin_um , inplace=True)
    print(f'AlgunaVez Tamaño dataframe Grupos : {len(dfgrupo)}')
    dfgrupo.to_csv(f'{os.getcwd()}\\result\\EI_AlgunaVez_2004_2013(GrupoLegalesIlegales).csv', index=False)
    # Sustancias
    list_cols, list_cols_sin_um = list_cols_Sust()
    dfsust.drop(columns=list_cols , inplace=True)
    print(f'AlgunaVez Tamaño dataframe Sustancias : {len(dfsust)}')
    dfsust.to_csv(f'{os.getcwd()}\\result\\EI_AlgunaVez_2004_2013(SustanciasLegalesIlegales).csv', index=False)
    return dfgrupo, dfsust

def UltimoMes(dfgrupo, dfsust):
    list_cols, list_cols_sin_um = list_cols_Grupo()
    dfgrupo.drop(columns=list_cols , inplace=True)
    print(f'UltimoMes Tamaño dataframe Grupos : {len(dfgrupo)}')
    dfgrupo.to_csv(f'{os.getcwd()}\\result\\EI_UltimoMes_2004_2013(GrupoLegalesIlegales).csv', index=False)
    # Sustancias
    list_cols, list_cols_sin_um = list_cols_Sust()
    dfsust.drop(columns=list_cols_sin_um , inplace=True)
    print(f'UltimoMes Tamaño dataframe Sustancias : {len(dfsust)}')
    dfsust.to_csv(f'{os.getcwd()}\\result\\EI_UltimoMes_2004_2013(SustanciasLegalesIlegales).csv', index=False)
    return dfgrupo, dfsust

def DrogaImpactoIlegalesGrupo(dfgrupo):
    list_cols, list_cols_sin_um = list_cols_Grupo()
    dfgrupo.drop(columns=list_cols + list_cols_sin_um, inplace=True)
    dummies = pd.get_dummies(dfgrupo['DrogaImpacto'])
    dfgrupo = pd.concat([dfgrupo, dummies], axis=1)
    print (f'DrogaImp Tamaño dataframe Grupos Ilegales : {len(dfgrupo)}')
    dfgrupo.to_csv(f'{os.getcwd()}\\result\\EI_DROGAIMP_2004_2013(GrupoIlegales).csv', index=False)

def DrogaImpactoIlegalesSustancias(dfsust):
    list_cols, list_cols_sin_um = list_cols_Sust()
    dfsust.drop(columns=list_cols + list_cols_sin_um, inplace=True)
    dummies = pd.get_dummies(dfsust['DrogaImpacto'])
    dfsust = pd.concat([dfsust, dummies], axis=1)
    print (f'DrogaImp Tamaño dataframe Sustancias Ilegales : {len(dfsust)}')
    dfsust.to_csv(f'{os.getcwd()}\\result\\EI_DROGAIMP_2004_2013(SustanciasIlegales).csv', index=False)

def AlgunaVezIlegalesGrupo(dfgrupo):
    list_cols, list_cols_sin_um = list_cols_Grupo()
    dfgrupo.drop(columns=list_cols_sin_um , inplace=True)
    print(f'AlgunaVez Tamaño dataframe Grupos Ilegales : {len(dfgrupo)}')
    dfgrupo.to_csv(f'{os.getcwd()}\\result\\EI_AlgunaVez_2004_2013(GrupoIlegales).csv', index=False)

def AlgunaVezIlegalesSust(dfsust):
    list_cols, list_cols_sin_um = list_cols_Sust()
    dfsust.drop(columns=list_cols , inplace=True)
    print(f'AlgunaVez Tamaño dataframe Sustancias Ilegales : {len(dfsust)}')
    dfsust.to_csv(f'{os.getcwd()}\\result\\EI_AlgunaVez_2004_2013(SustanciasIlegales).csv', index=False)

def UltimoMesIlegalesGrupo(dfgrupo):
    list_cols, list_cols_sin_um = list_cols_Grupo()
    dfgrupo.drop(columns=list_cols , inplace=True)
    print(f'UltimoMes Tamaño dataframe Grupos Ilegales : {len(dfgrupo)}')
    dfgrupo.to_csv(f'{os.getcwd()}\\result\\EI_UltimoMes_2004_2013(GrupoIlegales).csv', index=False)

def UltimoMesIlegalesSustancias(dfsust):
    list_cols, list_cols_sin_um = list_cols_Sust()
    dfsust.drop(columns=list_cols_sin_um , inplace=True)
    print(f'UltimoMes Tamaño dataframe Sustancias Ilegales : {len(dfsust)}')
    dfsust.to_csv(f'{os.getcwd()}\\result\\EI_UltimoMes_2004_2013(SustanciasIlegales).csv', index=False)
    return dfsust

def elim_EdadInicio(df):
    cols = [col for col in df.columns if 'EdadInicio' in col]
    df.drop(columns=cols, inplace=True)
    return df
```

```python
def main():
    dfgrupos, dfsust, dfGruposIlegales, dfSustanciasIlegales = Read_data(os.getcwd())
    dfgrupos.to_csv(f'{os.getcwd()}\\result\\EI_2004_2013(Grupo).csv', index=False)
    dfsust.to_csv(f'{os.getcwd()}\\result\\EI_2004_2013(Sustancias).csv', index=False)
    # # dfgrupos = elim_EdadInicio(dfgrupos)
    # # dfsust = elim_EdadInicio(dfsust)
    DrogaImpacto(dfgrupos.copy(), dfsust.copy())
    AlgunaVez(dfgrupos.copy(), dfsust.copy())
    UltimoMes(dfgrupos.copy(), dfsust.copy())
    DrogaImpactoIlegalesGrupo(dfGruposIlegales.copy())
    DrogaImpactoIlegalesSustancias(dfSustanciasIlegales.copy())
    AlgunaVezIlegalesGrupo(dfGruposIlegales.copy())
    AlgunaVezIlegalesSust(dfSustanciasIlegales.copy())
    UltimoMesIlegalesGrupo(dfGruposIlegales.copy())
    UltimoMesIlegalesSustancias(dfSustanciasIlegales.copy())
    return dfgrupos, dfsust, dfGruposIlegales, dfSustanciasIlegales
```
