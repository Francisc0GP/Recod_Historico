```python
import pandas as pd
import os
import re
```

```python
def asegurar_columnas_base(df, columnas_base):
    """
    Asegura que el DF tenga las columnas con los nombres en `columnas_base`.
    Si no existen, intenta renombrar <columna>UM -> <columna>.
    Lanza KeyError si aún faltan columnas después del intento.
    """
    # 1) ¿Faltan columnas base?
    faltantes = [c for c in columnas_base if c not in df.columns]

    if faltantes:
        # 2) Intentar renombrar las que existan como <base>UM -> <base>
        ren = {f"{c}UM": c for c in faltantes if f"{c}UM" in df.columns}
        if ren:
            df.rename(columns=ren, inplace=True)

        # 3) Verificar nuevamente
        faltantes = [c for c in columnas_base if c not in df.columns]
        if faltantes:
            raise KeyError(f"Siguen faltando columnas: {faltantes}")

    return df

def Read_data(file_path, filepath2):
    columnas_a_convertir = ['Tabaco', 'Alcohol', 'Marihuana', 'Hachis',	'Cocaína', 'Crack', 'Otras Presentaciones (Basuco o pasta base, cocaina negra)',	'Solventes y removedores',	'Pegamento',	'Esmaltes y pinturas', 	'Otros (aire comprimido, gasolinas y combustibles)', 	'Anfetaminas',	'Metanfetaminas',	'MDMA(extasis) y metanfetaminas alucinogenas (DMT)',	'Otros (derivados anfetaminicos)',	'LSD',	'Plantas alucinogenas y derivados',	'Otras (PCP, ketamina, excepto metanfetamina)',	'Benzodiazepinas',	'Rohypnol',	'Otras (sedantes hiptnoticos, GHB)',	'Heroina',	'Opiaceos sinteticos (propoxifeno, nailbufina)',	'Opio y opiodes (morfina, codeina)',	'Con utilidad medica (Prozac, Paxil, Carbamazepina)',	'Otras Sustancias']
    df = pd.read_csv(file_path)
    print (f'CIJ ECERIECS Tamaño del DataFrame después de leer el archivo: {df.shape} Nombre del archivo: {file_path}')
    df.drop(df[df['Año'] < 2014].index, inplace=True)
    df.drop(df[df['Año'] > 2020].index, inplace=True)
    print (f'CIJ ECERIECS Tamaño del DataFrame después de filtrar por año: {df.shape} Nombre del archivo: {file_path}')
    df2 = pd.read_csv(filepath2)
    df2.drop(columns=['DrogaImpacto'], inplace=True)
    df2 = asegurar_columnas_base(df2, columnas_a_convertir)
    df2[columnas_a_convertir] = df2[columnas_a_convertir].fillna(0)
    df2[columnas_a_convertir] = df2[columnas_a_convertir].replace({1: True, 0: False})
    df2 = df2.rename(columns={"Cocaína": "Cocaina", 'Sexo': 'SexoId'})
    return df, df2

def concat_data(df1, df2):
    df = pd.concat([df1, df2],axis = 0, ignore_index=True)
    return df

def main():
    list_files_2014 = ['\\data\\ECERIECS_AlgunaVez.csv', '\\data\\ECERIECS(SI)_AlgunaVez.csv', '\\data\\ECERIECS_DrogaImpacto.csv', '\\data\\ECERIECS(SI)_DrogaImpacto.csv', '\\data\\ECERIECS_UltimoMes.csv', '\\data\\ECERIECS(SI)_UltimoMes.csv']
    list_files_2004 = ['\\data\\EI_AlgunaVez_2004_2013(SustanciasLegalesIlegales).csv','\\data\\EI_AlgunaVez_2004_2013(SustanciasIlegales).csv', '\\data\\EI_DROGAIMP_2004_2013(SustanciasLegalesIlegales).csv', '\\data\\EI_DROGAIMP_2004_2013(SustanciasIlegales).csv', '\\data\\EI_UltimoMes_2004_2013(SustanciasLegalesIlegales).csv', '\\data\\EI_UltimoMes_2004_2013(SustanciasIlegales).csv' ]
    for (col,col2) in zip(list_files_2014, list_files_2004):
        df1, df2 = Read_data(f'{os.getcwd()}{col}', f'{os.getcwd()}{col2}')
        df = concat_data(df1, df2)
        nombre = re.search(r'[^\\]+$', col).group()
        print(nombre)
        print (f'Tamaño del DataFrame después de concatenar: {df.shape}')
        df.to_csv(f'{os.getcwd()}\\result\\{nombre}', index=False)
```

```python
main()
```
