```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
def Read_data():
    grupos_eceriecs = pd.read_csv(r'C:\Users\franc\Documents\OneDrive\Documentos\GitHub\Recod_Historico\NewFormat\data\GruposLegalesIlegales_ECERIECS_mod(EdadInicio).csv')
    grupos_pasado = pd.read_csv(r'C:\Users\franc\Documents\OneDrive\Documentos\GitHub\Recod_Historico\NewFormat\data\EI_2004_2013(Grupo).csv')
    sust_eceriecs = pd.read_csv(r'C:\Users\franc\Documents\OneDrive\Documentos\GitHub\Recod_Historico\NewFormat\data\ECERIECSLegalesIlegales_mod(EdadInicio).csv')
    sust_pasado = pd.read_csv(r'C:\Users\franc\Documents\OneDrive\Documentos\GitHub\Recod_Historico\NewFormat\data\EI_2004_2013(Sustancias).csv')
    return grupos_eceriecs, grupos_pasado, sust_eceriecs, sust_pasado

def rename_groups(grupos_eceriecs, grupos_pasado):
    grupos_pasado.rename(columns={'Sexo': 'SexoId'}, inplace=True)
    grupos_eceriecs = grupos_eceriecs.drop(columns=['EdadInicioTodasSustancias'])
    grupos_eceriecs = grupos_eceriecs[grupos_eceriecs['Año'].between(2014, 2020)]
    return grupos_eceriecs, grupos_pasado

def rename_sust (sust_eceriecs, sust_pasado):
    sust_pasado.rename(columns={'Sexo': 'SexoId'}, inplace=True)
    sust_eceriecs = sust_eceriecs.rename(columns={col: col.replace('UltimoMes', 'UM') for col in sust_eceriecs.columns})
    sust_eceriecs.rename(columns={'Cocaina': 'Cocaína', 'CocainaUM': 'CocaínaUM', 'EdadInicioCocaina': 'EdadInicioCocaína'}, inplace=True)
    sust_eceriecs = sust_eceriecs[sust_eceriecs['Año'].between(2014, 2020)]
    return sust_eceriecs, sust_pasado
```

```python
def main():
    grupos_eceriecs, grupos_pasado, sust_eceriecs, sust_pasado = Read_data()
    grupos_eceriecs, grupos_pasado = rename_groups(grupos_eceriecs, grupos_pasado)
    grupos_concat = pd.concat([grupos_eceriecs, grupos_pasado], ignore_index=True)
    sust_eceriecs, sust_pasado = rename_sust(sust_eceriecs, sust_pasado)
    sust_concat = pd.concat([sust_eceriecs, sust_pasado], ignore_index=True)
    return grupos_concat, sust_concat
```

```python
grupos_concat, sust_concat = main()
```

```python
grupos_concat.to_csv(r'C:\Users\franc\Documents\OneDrive\Documentos\GitHub\Recod_Historico\NewFormat\results\gruposHistoric_concat.csv', index=False)
sust_concat.to_csv(r'C:\Users\franc\Documents\OneDrive\Documentos\GitHub\Recod_Historico\NewFormat\results\sustHistoric_concat.csv', index=False)
```
